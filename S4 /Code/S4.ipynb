## Regresion  por cuantiles

```python
model = "lnwage ~ female+ educ + exper + expersq + tenure + married + divorced"
m1_10 = pf.quantreg(model, data, quantile = 0.1)
m1_25 = pf.quantreg(model, data, quantile = 0.25)
m1_50 = pf.quantreg(model, data, quantile = 0.5)
m1_75 = pf.quantreg(model, data, quantile = 0.75)
m1_90 = pf.quantreg(model, data, quantile = 0.9)
```

```python
pf.etable([m1_10,m1_25, m1_50, m1_75, m1_90] ,
          keep=['female']
         )
```

```python
import statsmodels.api as sm

# Submuestras
df_m = data[data['female'] == 0]
df_f = data[data['female'] == 1]

# Variables dependientes y regresores
y_m = df_m['lnwage']
X_m = sm.add_constant(df_m[['educ','exper','expersq','tenure','married','divorced']])

y_f = df_f['lnwage']
X_f = sm.add_constant(df_f[['educ','exper','expersq','tenure','married','divorced']])

# OLS con errores robustos
model_m = sm.OLS(y_m, X_m).fit(cov_type='HC1')
model_f = sm.OLS(y_f, X_f).fit(cov_type='HC1')

# Medias de las covariables
mean_X_m = X_m.mean()
mean_X_f = X_f.mean()

# Diferencia promedio de lnwage
gap = y_m.mean() - y_f.mean()

# Efecto explicado (endowments)
explained = (mean_X_m - mean_X_f) @ model_f.params

# Efecto no explicado (coefficients)
unexplained = mean_X_m @ (model_m.params - model_f.params)

print(f"Gap: {gap:.4f}, Explained: {explained:.4f}, Unexplained: {unexplained:.4f}")

```
